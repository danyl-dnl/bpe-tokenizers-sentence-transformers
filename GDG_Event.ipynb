```python
def ecode_text(text_arr, encoder):
 rows = list()


 for txt in text_arr:
   ids = encoder.encode(txt)
   piece = [encoder.decode([id]) for id in ids]
   rows.append({
       "text": txt,
       "ids": ids,
       "pieces": piece,
       "count": len(ids)
   })


 return rows
```

```python
pd.DataFrame(ecode_text(test_str, gp2_encoder))
```

```python
from collections import Counter
```

```python
corpus = [
   "low",
   "lower",
   "lowest",
   "new",
   "newer",
   "newest",
   "wide",
   "wider",
   "widest",
]


words = [list(word) + ["</w>"] for word in corpus]
words[:3]
```

```python
def pair_count(words):
  ctr = Counter()
  for word in words:
    for w1,w2 in zip(word, word[1:]):
      ctr[(w1,w2)] += 1
  return ctr

pair_count(words)
```

```python
def merge_pair(words, pair):
    new_words = []

    for word in words:
        new_word = []
        idx = 0

        while idx < len(word):
            if (
                idx < len(word) - 1
                and (word[idx], word[idx + 1]) == pair
            ):
                new_word.append(word[idx] + word[idx + 1])
                idx += 2
            else:
                new_word.append(word[idx])
                idx += 1

        new_words.append(new_word)

    return new_words
```

```python
pair_cnt = pair_count(words)
pair_cnt.most_common(1)
```

```python
merge_history = list()
for step in range(5):
  pair_counts = pair_count(words)
  best_pair, count = pair_counts.most_common(1)[0]

  print(f"Step {step + 1}: merge {best_pair} count {count}")
  merge_history.append((best_pair,count))
  words = merge_pair(words,best_pair)
```

```python
from sentence_transformers import SentenceTransformer


sentence_model = SentenceTransformer(
   "sentence-transformers/all-MiniLM-L6-v2"
)


sentence_examples = [
   "My laptop battery is dying quickly.",
   "The computer does not last long without charging.",
   "I love playing football.",
   "My notebook loses power too fast.",
]


sentence_vectors = sentence_model.encode(
   sentence_examples,
   normalize_embeddings=True,
)


print("Embedding shape:", sentence_vectors.shape)
```
